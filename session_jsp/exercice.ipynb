```python
cluster_size = 4096
import math
def get_my_size(p):
    if p.is_symlink():
        return 0
    elif p.is_file():
        return math.ceil(p.stat().st_size / cluster_size) * cluster_size
    elif p.is_dir():
        return sum(get_my_size(child) for child in p.iterdir())
    else:
        return 0
```

```python
import os
from pathlib import Path
def get_folder_size(p):
    if(p.is_file()):
        print(f"{p} : {p.stat().st_size} bytes")
        return p.stat().st_size
    if p.is_dir():        
        return sum(get_folder_size(f) for f in p.iterdir())
    return 0

def bytes_to_mB(bytes):
    return str(bytes / (1024 * 1024)) + "Mo"
    
```
