## Exercise 3: Counting

```python
count = 0
for x in data:
    if x > 0:
        count += 1
```

```python
count = sum(1 for x in data if x > 0)
```

## Exercise 5: Building a list

```python
result = []
for x in numbers:
    if x % 2 == 0:
        result.append(x * 2)
```

```python
result = [x * 2 for x in numbers if x % 2 == 0]
```

## Exercise 13: Nested loops

```python
pairs = []
for i in range(len(a)):
    for j in range(len(b)):
        pairs.append((a[i], b[j]))
```

```python
pairs = [(x, y) for x in a for y in b]
```

## Exercise 14: Flattening a list

```python
flat = []
for sublist in data:
    for x in sublist:
        flat.append(x)
```

```python
flat = [x for sublist in data for x in sublist]
```

## Exercise 15: First match

```python
result = None
for x in data:
    if x > 10:
        result = x
        break
```

```python
result = next((x for x in data if x > 10), None)
```

## Exercise 17: Remove duplicates
